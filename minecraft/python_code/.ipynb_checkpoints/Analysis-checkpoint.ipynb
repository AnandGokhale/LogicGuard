```python
# List all files in the directory
def construct_df_list(directory):
    df_list = []
    for filename in os.listdir(directory):
        if filename.endswith("_full_state.pkl"):
            filepath = os.path.join(directory, filename)
            try:
                with open(filepath, 'rb') as f:
                    data = pickle.load(f)
    
    
                rows = []
    
                if isinstance(data, list):       
                    print(f"File: {filename} → Length of list: {len(data)}")
                    for i, entry in enumerate(data):
    
                        state = entry['state']
    
                        row = {
                            "file": filename,
                            "step": i,
                            "time": entry.get('time', None),
                            "obs_props": get_obs_props(state),
                            "critic": state.critic,
                            "previous_code": state.codeLastRound,
                            "previous_error": state.executionError if state.executionError else "None",
                            "previous_output": state.Output if state.Output else "None",
                            "position": [state.position.x,state.position.y,state.position.z],
                            "hand": state.equipment.hand,
                            "inventory": state.inventory,
                            "nearbyBlocks": state.nearbyBlocks,
                            "code": entry['code'],
                            "action": get_active_keys(entry['action']),
                            "failed_codes": entry.get('failed_codes', None),
                            "confusion_matrix": entry.get('confusion_matrix', None),
                        }
    
                        rows.append(row)
                df = pd.DataFrame(rows)
                df_list.append(df)
                        
                        
    
    
            except Exception as e:
                print(f"Error reading {filename}: {e}")
    return df_list
```

```python
def preprocess_df(df):
    print(df.keys())
    df['time'] = df['time'] - df['time'].iloc[0]
    df['critic'] = df['critic'].apply(
        lambda x: [s for s in re.sub(r'^.*rules:\s*', '', x).split('\n') if s] if isinstance(x, str) else []
    )    
    df['num_failed_codes'] = df['failed_codes'].apply(len)
    code_heads = []
    code_reasons = []
    
    for code_list in df['failed_codes']:
        head_list = []
        reason_list = []
        
        for code in code_list:
            if "critic recommends that" in code:
                before, after = code.split("critic recommends that", 1)
                head = before.strip().split(";")[0] + ";"  # extract up to first semicolon
                reason = after.strip()
            else:
                head = code.strip().split(";")[0] + ";"
                reason = ""  # no critic explanation
                
            head_list.append(head)
            reason_list.append(reason)
        
        code_heads.append(head_list)
        code_reasons.append(reason_list)
    
    df['failed_code_head'] = code_heads
    df['failed_code_reason'] = code_reasons


```

```python
def first_true_occurrence(df, parameter):
    """
    Returns the first timestep index where df['obs_props'][parameter] is True.
    If the parameter never becomes True, returns None.
    """
    for i, row in df.iterrows():
        obs_props = row.get('obs_props', {})
        if obs_props.get(parameter, False):
            return i
    return None  # or len(df) if you want to assume failure as end of episode

```

```python
from collections import defaultdict
import numpy as np

def table_1_populator(df_list, obs_props):

    result = defaultdict(dict)

    for prop in obs_props:
        steps = []
        successes = 0

        for df in df_list:
            step = first_true_occurrence(df, prop)
            if step is not None:
                steps.append(step)
                successes += 1

        mean_step = np.mean(steps) if steps else None
        std_step = np.std(steps) if steps else None

        success_rate = successes / len(df_list)

        result[prop]['mean_step'] = mean_step
        result[prop]['std_step'] = std_step
        result[prop]['success_rate'] = success_rate

    return result
```

```python

def safety_violation_counter(df_list, column="previous_error"):
    """
    Count how often the `column` is not "None" across a list of DataFrames.
    
    Returns:
        summary_df : DataFrame with per-df counts and percentages
        aggregate  : dict with total rows, total errors, overall percentage
    """
    results = []
    
    total_rows = 0
    total_errors = 0
    error_rows_list = []

    for i, df in enumerate(df_list):
        row_count = len(df)
        error_count = (df[column] != "None").sum()

        total_rows += row_count
        total_errors += error_count
        error_rows_list.append(df[df[column] != "None"][column].to_string(index=False))


        results.append({
            "df_index": i,
            "rows": row_count,
            "errors": error_count,
            "error_pct": error_count / row_count if row_count > 0 else 0.0,
        })

    summary_df = pd.DataFrame(results)

    for e in error_rows_list:
        print(e)

    aggregate = {
        "total_rows": total_rows,
        "total_errors": total_errors,
        "aggregate_error_pct": total_errors / total_rows if total_rows > 0 else 0.0
    }

    print(aggregate)

    return summary_df, aggregate
```

```python
def analyze_confusion(confusion_matrix):
    print("Confusion Matrix Analysis:\n", confusion_matrix)
    confusion_matrix = confusion_matrix/ np.sum(confusion_matrix)
    print("Normalized Confusion Matrix:\n", confusion_matrix)
    print("FN + TP", confusion_matrix[0,0] + confusion_matrix[1,1])
    print("FP + TN", confusion_matrix[1,0] + confusion_matrix[0,1])


def summarize_confusion_matrices(df_list, column="confusion_matrix"):
    total_cm = np.zeros((2, 2), dtype=int)
    for df in df_list:  
        for _, row in df.iterrows():
            cm = np.array(row['confusion_matrix'], dtype=int)
            total_cm += cm


    analyze_confusion(total_cm)
    return
```

```python
def failed_code_frequencies(df_list, column_name="failed_code_reason"):
    per_df_counts = {}
    total_counter = Counter()
    total_rows = 0
    rows_with_failures = 0

    for i, df in enumerate(df_list):
        total_rows += len(df)

        # Count rows with non-empty failed_code_reason
        df_non_empty = df[df[column_name].apply(lambda x: isinstance(x, list) and len(x) > 0)]
        rows_with_failures += len(df_non_empty)

        # Flatten list of lists
        all_failed = sum(df_non_empty[column_name], [])
        freq = Counter(all_failed)

        # Store and update
        per_df_counts[f"df_{i}"] = freq
        total_counter.update(freq)

    # Compute failure percentage
    failure_percentage = (rows_with_failures / total_rows) * 100 if total_rows > 0 else 0.0

    return per_df_counts, total_counter, failure_percentage
```

```python
per_df_counts, total_failed_counts, failure_percentage = failed_code_frequencies(df_list_critic)

# Compute average number of failures per row
total_rows = sum(len(df) for df in df_list)
total_failures = sum(total_failed_counts.values())
num_dfs = len(df_list)

# Now calculate: (total_rows + total_failures) / num_dfs
combined_stat = (total_rows + total_failures) / num_dfs

print(f"Total rows: {total_rows}")
print(f"Total failure entries: {total_failures}")
print(f"Total Failure Percentage: {failure_percentage}")
print(f"Average (total_rows + total_failures) / num_dfs: {combined_stat:.2f}")
```

```python
from collections import defaultdict


def error_frequencies_with_codes(df_list, error_col="previous_error", code_col="previous_code"):
    per_df_counts = {}
    combined_counts = defaultdict(int)
    combined_code_examples = defaultdict(list)

    total_rows = 0
    total_errors = 0

    for i, df in enumerate(df_list):
        total_rows += len(df)
        
        # Filter out actual NaNs and string "None"
        filtered_df = df[(df[error_col].notna()) & (df[error_col] != "None")]
        total_errors += len(filtered_df)

        # Count frequency
        error_counts = filtered_df[error_col].value_counts()
        per_df_counts[f"df_{i}"] = error_counts.sort_values(ascending=False)

        # Update combined stats
        for _, row in filtered_df.iterrows():
            err = row[error_col]
            code = row[code_col]
            combined_counts[err] += 1
            if len(combined_code_examples[err]) < 5:  # cap at 5 examples
                combined_code_examples[err].append(code)

    # Final result
    combined_series = pd.Series(combined_counts).sort_values(ascending=False)
    error_percentage = (total_errors / total_rows) * 100 if total_rows > 0 else 0.0

    return per_df_counts, combined_series, combined_code_examples, error_percentage
```
